```python
# Estrutura de decisões

# Entrada

numero1 = int (input ("digite um número: "))
numero2 = int (input ("digite um número: "))

# Procesamento e Saída

if numero1 > numero2:
  print(f"o numero {numero1} é maior que o {numero2}")

elif numero1 == numero2:
  print(f"o numero {numero1} é igual ao {numero2}")

else:
  print(f"o numero {numero1} é menor que o numero {numero2}")

```

```python
# Escrever um pograma onde o usuario digita o nome e a idade:

# identificar se o usuario é criança, adolescente ou adulto

# > 12 Criança
# > 18 adoslecente
# < 18 adulto

nome = str(input("digite seu nome: "))
idade = int(input("digite sua idade: "))

if idade <= 12:
    print(f"Olá, {nome}, você é criança")

elif idade <= 18:
    print(f"Olá, {nome}, você é adoslecente")

else:
  print(f"Olá,{nome}, você é adulto")
```
