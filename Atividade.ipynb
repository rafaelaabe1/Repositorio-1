```python
# Classificação de idade

# > 17 menor de idade
# > 18 a 59 adulto
# < 60 idoso

idade = int(input("digite sua idade: "))

if idade <= 17:
    print(f"Você é menor de idade")

elif idade >= 18 and idade <= 59 :
    print(f"Você é adulto")

else:
  print(f"Você é idoso")
```

```python
# Cálculo de desconto

valor = float (input("Digite o valor da compra:"))

desconto1 = 10
calculo1 = valor * (desconto1 / 100)
calculo2 = valor - calculo1

desconto2 = 20
calculo3 = valor * (desconto2 / 100)
calculo4 = valor - calculo3

if valor <= 100:
  print(f"sem desconto")

elif valor <= 500:
  print(f"você tem direito de um desconto de 10%, valor final é de R${calculo2}")

else:
  print(f"você tem direito de um desconto de 20%, valor final é de R${calculo4}")
```

```python
#triangulo

l1= int(input("digite um lado do triangulo: "))
l2= int(input("digite um lado do triangulo: "))
l3= int(input("digite um lado do triangulo: "))

if (l1+l2>l3) and (l2+l3>l1) and (l3+l1>l2):
  print("é um triangulo.")

else:
  print("nao é um triangulo")
```

```python
# Sistema energia

energia = int(input("digite a energia do personagem: "))

if energia <= 0:
  print("morto")

elif energia >= 1 and energia <= 30:
  print("critico")

elif energia >= 31 and energia <= 70:
  print("normal")

else:
  print("saudavel")
```

```python
#media escolar

media = int(input("digite a media final: "))

if media >= 7:
  print("aprovado")

elif media <= 6.9:
  print("recuperação")

else:
  print("reprovado")
```
