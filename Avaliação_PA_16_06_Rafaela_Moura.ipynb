```python
# Contagem Crescente

#Elabore um programa que exiba na tela os números de 1 até 20, um em cada linha.

contador = 1

while contador <= 20:
  print(contador)
  contador += 1
```

```python
# Soma dos Números

num = int (input("digite um numero: "))
soma = 0

for i in range(1, num + 1):
  soma += i

print(f"resultado: {soma}")
```

```python
# Tabuada

# Solicite ao usuário um número inteiro e exiba sua tabuada de 1 até 10

n = int (input("Digite um número: "))

for i in range(1, 11):
    resultado = n * i
    print(f"{n} x {i} = {resultado}")
```

```python
# Fatorial

# Solicite ao usuário um número inteiro positivo e calcule seu fatorial utilizando um laço de repetição

while True:
    try:
        numero = int(input("Digite um número: "))
        if numero < 0:
            print("Digite um número positivo.")
        else:
            break
    except ValueError:
        print("Digite um número inteiro.")

fatorial = 1
if numero == 0:
    print(f"O fatorial de {numero} é 1")
elif numero > 0:
    for i in range(1, numero + 1):
        fatorial *= i
    print(f"O fatorial de {numero} é {fatorial}")
```

```python
# Caixa Eletrônico

#Solicite ao usuário o valor do saque e informe a menor quantidade possível de notas para realizar o pagamento

# Um caixa eletrônico possui notas de:
#R$ 100
#R$ 50
#R$ 20
#R$ 10
#R$ 5
#R$ 2
#R$ 1

valor = int (input("Digite o valor que quer sacar: "))

valor_rest = valor

for nota in (100, 50, 20, 10, 5, 2, 1):

    quantidade_notas = valor_rest // nota

    if quantidade_notas > 0:
        print(f"{quantidade_notas} nota(s) de R$ {nota}")
    valor_rest %= nota
```
