```python
# Par ou ímpar

# Entrada

numero = int (input ("digite um número: "))

# Procesamento e Saída

if numero % 2 == 0:
  print(f"o numero é par")

else:
  print(f"o numero é ímpar")
```

```python
# Maior de dois números

# Entrada

numero1 = int (input ("digite um número: "))
numero2 = int (input ("digite um número: "))

# Procesamento e Saída

if numero1 > numero2:
  print(f"o numero {numero1} é maior que o {numero2}")

else:
  print(f"o numero {numero1} é igual ao {numero2}")
```

```python
# Aprovação de aluno

#	Aprovado (nota ≥ 7)
#	Recuperação (5 ≤ nota < 7)
#	Reprovado (nota < 5)

media = int(input("digite sua nota: "))

if media >= 7:
  print("aprovado")

elif media >= 5:
  print("recuperação")


else:
  print("reprovado")
```

```python
# Classificação de idade

#	Criança (0–12)
#	Adolescente (13–17)
#	Adulto (18–59)
#	Idoso (60+)

idade = int(input("digite sua idade: "))

if idade <= 12:
    print(f"Você é criança")

elif idade >= 13 and idade <= 17:
    print(f"Você é adolescente")

elif idade >= 18 and idade <= 59 :
    print(f"Você é adulto")

else:
  print(f"Você é idoso")
```

```python
# Senha correta

senha = str(input("Digite sua senha: "))

if senha == "senha67":
  print("senha correta")

else:
  print("senha incorreta")
```

```python
# Positivo, negativo ou zero

numero = int (input ("digite um número: "))

if numero > 0:
  print("numero positivo")

elif numero < 0:
  print("numero negativo")

else:
  print("zero")
```

```python
# Calculadora simples

numero1 = float(input("digite um numero: "))
numero2 = float(input("digite um numero: "))
operacao = input("Digite a operação (+, -, *, /): ")

mais = numero1 + numero2
menos = numero1 - numero2
vezes = numero1 * numero2
divisao = numero1 / numero2

if operacao == '+':
  print(f"resultado igual a: {mais}")

elif  operacao == '-':
  print(f"resultado igual a: {menos}")

elif  operacao == '*':
  print(f"resultado igual a: {vezes}")

elif  operacao == '/':
  print(f"resultado igual a: {divisao}")
```

```python
# Dia da semana

semana = int(input("Digite um dia da semana:"))

if semana == 1:
  print("Segunda")

elif semana == 2:
  print("terça")

elif semana == 3:
  print("quarta")

elif semana == 4:
  print("quinta")

elif semana == 5:
  print("sexta")

elif semana == 6:
  print("sabado")

elif semana == 7:
  print("domingo")
```

```python
# Ano bissexto

ano = int(input("Digite um ano: "))

if (ano % 400 == 0) or (ano % 4 == 0 and ano % 100 != 0):
    print(f"{ano} é um ano bissexto.")
else:
    print(f"{ano} não é um ano bissexto.")
```

```python
#	Desconto em compras

valor = float (input("Digite o valor da compra:"))

desconto1 = 10
calculo1 = valor * (desconto1 / 100)
calculo2 = valor - calculo1

desconto2 = 5
calculo3 = valor * (desconto2 / 100)
calculo4 = valor - calculo3

if valor >= 100:
  print(f"você tem direito a um desconto de 10%, valor final é de R${calculo2}")

elif valor >= 50 and valor <= 100:
  print(f"você tem direito a um desconto de 5%, valor final é de R${calculo4}")

else:
  print(f"Sem desconto")
```
