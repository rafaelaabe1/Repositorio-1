```python
# Operadores aritimeticos

# Pedir ao usuario para digitar 02 numeros

# Entrada

a = int(input("digite um numero: "))
b = int(input("digite um outro numero: "))

#processamento

soma1 = a + b
soma2 = a - b
soma3 = a * b
soma4 = a / b
div_int = a // b
soma6 = a % b
soma7 = a ** b

#saida

print("----------OPERADORES ARITIMETICOS----------")
print("")
print("----------SOMA----------")
print(f"a soma de {a} mais {b} é igual a: {soma1} ")
print("")
print("----------SUBTRAÇÃO----------")
print(f"a SUBTRAÇÃO de {a} MENOS {b} é igual a: {soma2} ")
print("")
print("----------MULTIPLICAÇÃO----------")
print(f"a MULTIPLICAÇÃO de {a} POR {b} é igual a: {soma3}")
print("")
print("----------DIVISÃO----------")
print(f"a DIVISÃO de {a} POR {b} é igual a: {soma4}")
print("")
print("----------DIVISÃO INTEIRA----------")
print(f"a DIVISÃO de {a} POR {b} é igual a: {div_int}")
print("")
print("----------RESTO----------")
print(f"o RESTO de {a} POR {b} é igual a: {soma6}")
print("")
print("----------POTENCIA----------")
print(f"a POTENCIA de {a} POR {b} é igual a: {soma7}")
print("")
print("-----------------------------------------------")
```
